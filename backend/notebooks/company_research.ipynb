```python
import pandas as pd

# Load dataset
df = pd.read_excel("datasets/million_company_names_with_serial.xlsx")

# Display first rows
df.head()
```

```python
# Clean column names
df.columns = [col.strip() for col in df.columns]

# Preview cleaned dataset
df.head()
```

```python
# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.head()
```

```python
df["company_name"]
```

```python
# Search company function

def search_company(company_name):
    results = df[df['company_name'].str.contains(company_name, case=False, na=False)]
    return results.head(10)

# Example
search_company("Tech")
```
