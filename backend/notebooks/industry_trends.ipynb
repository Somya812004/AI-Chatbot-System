# Industry Trends Analysis
Monitoring emerging technologies and market shifts in the tech industry using PathForge AI.

```python
import os
import pandas as pd
import requests
import matplotlib.pyplot as plt
import seaborn as sns
from dotenv import load_dotenv

# Setup environment
load_dotenv('../.env')
BASE_URL = "http://127.0.0.1:5000/api/research"

print('Industry Trends workspace ready.')
```

## 1. Fetching Current Market Trends
Retrieving the latest technology trends from the research module.

```python
response = requests.get(f"{BASE_URL}/trends")
trends_data = response.json()

df_trends = pd.DataFrame(trends_data)
print("Fetched Trends:")
display(df_trends)
```

## 2. Visualizing Demand Shifts
Analyzing the growth potential of various domains.

```python
# Sample data based on current market reports for visualization
visual_data = {
    'Domain': ['Generative AI', 'Cybersecurity', 'Cloud Native', 'Sustainability Tech', 'Quantum Computing'],
    'Growth_Rate (%)': [300, 150, 120, 80, 45]
}
df_viz = pd.DataFrame(visual_data)

plt.figure(figsize=(10, 6))
sns.barplot(x='Growth_Rate (%)', y='Domain', data=df_viz, palette='viridis')
plt.title('Industry Demand Growth Rate (2024-2025)')
plt.show()
```

## 3. AI-Powered Strategic Guidance
Consulting the PathForge AI research engine for deep insights on a specific trend.

```python
def get_ai_insight(query):
    payload = {"query": query}
    res = requests.post(f"{BASE_URL}/ai-guidance", json=payload)
    return res.json().get('response', 'No response available')

trend_query = "Explain the future scope of Generative AI in the financial sector."
insight = get_ai_insight(trend_query)

print(f"--- AI INSIGHT: {trend_query} ---\n")
print(insight)
```
