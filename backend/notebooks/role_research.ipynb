# Role Research & Analysis
Deep dive into job roles, skill requirements, and career roadmaps using PathForge AI.

```python
import os
import pandas as pd
import requests
import matplotlib.pyplot as plt
from dotenv import load_dotenv

# Setup environment
load_dotenv('../.env')
BASE_URL = "http://127.0.0.1:5000/api/research"

print('Role Research workspace ready.')
```

## 1. Exploring Roles
Fetching a list of available roles and their current demand levels.

```python
response = requests.get(f"{BASE_URL}/roles")
roles_data = response.json()

df_roles = pd.DataFrame(roles_data)
print("Available Roles & Demand:")
display(df_roles)
```

## 2. Generating AI Career Roadmaps
Requesting a structured learning path for a specific target role.

```python
target_role = "AI Engineer"
roadmap_res = requests.get(f"{BASE_URL}/roadmaps", params={"role": target_role})
roadmap_data = roadmap_res.json()

print(f"--- CAREER ROADMAP FOR: {target_role} ---\n")
print(roadmap_data.get('roadmap', 'Roadmap not found'))
```

## 3. Role-Based AI Guidance
Using the specialized PathForge research prompt for deep role insights.

```python
def get_role_insight(role):
    query = f"Explain the daily responsibilities and required projects for a {role}."
    payload = {"query": query}
    res = requests.post(f"{BASE_URL}/ai-guidance", json=payload)
    return res.json().get('response', 'No response available')

detailed_insight = get_role_insight(target_role)
print(f"--- DEEP DIVE: {target_role} ---\n")
print(detailed_insight)
```
