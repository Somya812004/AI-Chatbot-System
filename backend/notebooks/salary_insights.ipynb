# Salary Insights & Benchmarking
Comprehensive analysis of compensation trends using PathForge AI and live market data.

```python
import os
import pandas as pd
import requests
import matplotlib.pyplot as plt
import seaborn as sns
from dotenv import load_dotenv

# Setup environment
load_dotenv('../.env')
BASE_URL = "http://127.0.0.1:5000/api/research"

print('Salary Insights workspace ready.')
```

## 1. Structured Salary Benchmarks
Fetching standardized salary data for various tech roles.

```python
roles_to_check = ["Software Engineer", "AI Engineer", "Data Scientist", "Cybersecurity Analyst"]
salary_list = []

for role in roles_to_check:
    res = requests.get(f"{BASE_URL}/salary", params={"role": role})
    salary_list.append(res.json())

df_salary = pd.DataFrame(salary_list)
print("Salary Benchmarks:")
display(df_salary)
```

## 2. Visualizing Compensation Trends
Comparing average salaries across different domains.

```python
# Data cleaning for visualization
df_salary['avg_numeric'] = df_salary['avg_salary'].str.replace('₹', '').str.replace(',', '').astype(float)

plt.figure(figsize=(10, 6))
sns.barplot(x='avg_numeric', y='role', data=df_salary, palette='magma')
plt.title('Average Salary Comparison (INR)')
plt.xlabel('Salary (₹)')
plt.show()
```

## 3. AI-Powered Salary Negotiations & Strategy
Consulting PathForge AI for specialized advice on salary growth and negotiation.

```python
def get_salary_advice(role, experience):
    query = f"Provide a detailed salary growth strategy for a {role} with {experience} years of experience. Mention high-paying companies."
    payload = {"query": query}
    res = requests.post(f"{BASE_URL}/ai-guidance", json=payload)
    return res.json().get('response', 'No response available')

negotiation_insight = get_salary_advice("AI Engineer", "2")
print("--- AI STRATEGY REPORT ---\n")
print(negotiation_insight)
```
