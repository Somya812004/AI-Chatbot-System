```python
resume = {}
```

```python
def ask(question, key):
    ans = input(question + "\n> ")
    resume[key] = ans.strip()
    print("Got it\n")
```

```python
ask("What is your Full Name?", "name")
ask("What is your Professional Title?", "title")
ask("Enter your Phone Number:", "phone")
ask("Enter your Email:", "email")
ask("Enter your LinkedIn:", "linkedin")
ask("Enter your GitHub (or type N/A):", "github")
ask("Enter your Location (City, Country):", "location")

ask("Write a short Professional Summary (2 lines):", "summary")

ask("Enter Company Name:", "company")
ask("Enter Role:", "role")
ask("Enter Duration (e.g., Jan 2024 – Present):", "duration")
```

```python
# Projects
projects = []
for i in range(2):
    print(f"\nProject {i+1}")
    name = input("Project Name: ")
    tech = input("Tech Stack: ")
    date = input("Dates: ")

    print("Enter 3 bullet points:")
    p1 = input("• ")
    p2 = input("• ")
    p3 = input("• ")

    projects.append({
        "name": name,
        "tech": tech,
        "date": date,
        "points": [p1, p2, p3]
    })

resume["projects"] = projects
```

```python
ask("Enter College Name:", "college")
ask("Enter Degree:", "degree")
ask("Enter Graduation Year:", "grad")

ask("Enter Technical Skills (comma-separated):", "skills")
ask("Enter Soft Skills:", "soft")
ask("Enter Achievements:", "achievements")
```

```python
# Generate Resume
print("\n\n========== FINAL RESUME ==========\n")

print(resume["name"].upper())
print(resume["title"])
print(f'{resume["phone"]} | {resume["email"]} | {resume["linkedin"]} | {resume["github"]} | {resume["location"]}\n')

print("PROFESSIONAL SUMMARY")
print(resume["summary"], "\n")

print("EXPERIENCE")
print(f'{resume["company"]} | {resume["role"]} | {resume["duration"]}')
for pt in resume["experience"]:
    print(f'• {pt}')
print()

print("PROJECTS")
for p in resume["projects"]:
    print(f'{p["name"]} | {p["tech"]} | {p["date"]}')
    for pt in p["points"]:
        print(f'• {pt}')
    print()

print("EDUCATION")
print(f'{resume["college"]} | {resume["degree"]} | {resume["grad"]}\n')

print("SKILLS")
print(resume["skills"], "|", resume["soft"], "\n")

print("ACHIEVEMENTS")
print(resume["achievements"])
```
